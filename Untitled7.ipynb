```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. إعداد البيانات
datagen = ImageDataGenerator(rescale=1./255, validation_split=0.2)

# 2. تحميل البيانات (تأكدي من تغيير مسار '/content/dataset' لمسار ملفاتك)
train_data = datagen.flow_from_directory(
    '/content/dataset', target_size=(150, 150), batch_size=32, class_mode='binary', subset='training')

val_data = datagen.flow_from_directory(
    '/content/dataset', target_size=(150, 150), batch_size=32, class_mode='binary', subset='validation')
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, BatchNormalization, Dropout
from tensorflow.keras.optimizers import Adam

# بناء هيكل النموذج
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(150,150,3)),
    BatchNormalization(), MaxPooling2D(2,2),
    Conv2D(64, (3,3), activation='relu'),
    BatchNormalization(), MaxPooling2D(2,2),
    Flatten(),
    Dense(128, activation='relu'), Dropout(0.3),
    Dense(1, activation='sigmoid')
])

# إعداد النموذج للعمل
model.compile(optimizer=Adam(learning_rate=0.001), loss='binary_crossentropy', metrics=['accuracy'])

# بدء التدريب
history = model.fit(train_data, epochs=10, validation_data=val_data)
```

```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models

# بناء نموذج MobileNetV2
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(150, 150, 3))
base_model.trainable = False

mobilenet_model = models.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid')
])

mobilenet_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# تدريب النموذج
history_mobilenet = mobilenet_model.fit(train_data, epochs=10, validation_data=val_data)
```

```python
plt.plot(history_mobilenet.history['loss'], label='Train Loss')
plt.plot(history_mobilenet.history['val_loss'], label='Validation Loss')
plt.title('MobileNetV2 Loss Graph')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

# المقارنة بين دقة التدريب (Accuracy)
plt.figure(figsize=(12, 5))

# رسم دقة CNN
plt.subplot(1, 2, 1)
plt.plot(history.history['val_accuracy'], label='CNN Accuracy', color='blue', linestyle='--')
plt.plot(history_mobilenet.history['val_accuracy'], label='MobileNetV2 Accuracy', color='red')
plt.title('Comparison: Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

# المقارنة بين خسارة التدريب (Loss)
plt.subplot(1, 2, 2)
plt.plot(history.history['val_loss'], label='CNN Loss', color='blue', linestyle='--')
plt.plot(history_mobilenet.history['val_loss'], label='MobileNetV2 Loss', color='red')
plt.title('Comparison: Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# تقييم النموذج على بيانات الاختبار
loss, accuracy = model.evaluate(val_data)

print(f"Validation Loss: {loss:.4f}")
print(f"Validation Accuracy: {accuracy*100:.2f}%")
```
